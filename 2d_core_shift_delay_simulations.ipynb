```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1.inset_locator import inset_axes, mark_inset
from astropy.io import fits
from matplotlib.colors import SymLogNorm

norm = SymLogNorm(linthresh=0.5, linscale=1, vmin=.05, vmax=15)

# --- load your data + build extents exactly as before ---
h86    = fits.open(fn_86GHz)[0]
data86 = h86.data.squeeze()
hdr86  = h86.header

h230    = fits.open(fn_230GHz)[0]
data230 = h230.data.squeeze()
hdr230  = h230.header

# pixel scales → µas/pix
deg2µas = 3600*1e6
px86 = hdr86["CDELT1"] * deg2µas, hdr86["CDELT2"] * deg2µas
px230 = hdr230["CDELT1"] * deg2µas, hdr230["CDELT2"] * deg2µas

# build “centered” extents
ny86, nx86 = data86.shape
ny230, nx230 = data230.shape

ext86 = [
    -0.5*nx86*px86[0], 0.5*nx86*px86[0],
    -0.5*ny86*px86[1], 0.5*ny86*px86[1],
]

ext230_nom = [
    -0.5*nx230*px230[0], 0.5*nx230*px230[0],
    -0.5*ny230*px230[1], 0.5*ny230*px230[1],
]

# apply an integer‐pixel shift to the 230GHz map
shift_x, shift_y = -3735, 3455

dx = shift_x*px230[0]
dy = shift_y*px230[1]

ext230 = [
    ext230_nom[0] + dx,
    ext230_nom[1] + dx,
    ext230_nom[2] + dy,
    ext230_nom[3] + dy,
]

y_shift_pixels = 90
x_shift_pixels = -100

data86 = np.roll(np.roll(data86, y_shift_pixels, axis=0), x_shift_pixels, axis=1)

```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1.anchored_artists import AnchoredSizeBar
from matplotlib.patches import Ellipse


plt.rcParams.update({
    "font.size"   : 32             # global font size
})
# choose a bar length in *pixels*
bar_px_parent = 100     # e.g. 100 pixels long on the main map
bar_px_inset  =  1    # e.g. 20 pixels long in the inset

# convert to µas
bar_len_parent = bar_px_parent * px86[0]
bar_len_inset  = bar_px_inset  * px86[0]

fig, ax = plt.subplots(figsize=(20,20), dpi=200)

im = ax.imshow(data86, aspect='auto', origin='lower', cmap='gray_r', norm=norm, extent=ext86)

ax.contour(data230,
           levels=[1e-4,3e-4,6e-4],
           colors="white",
           linewidths=1,
           origin="lower",
           extent=ext230)

cb = ax.figure.colorbar(im, ax=ax, orientation='vertical', pad=0.01)
cb.set_label(f"$S$ (Jy/beam)")

ax.set_xlabel("RA offset (mas)")
ax.set_ylabel("Dec offset (mas)")
ax.set_title("86 GHz + shifted 230 GHz contours")

# after plotting…
ticks = ax.get_xticks()             # grab the current tick *positions*
ax.set_xticks(ticks)                # (not strictly necessary if you’re keeping them)
ax.set_xticklabels(ticks / 1000)    # divide each tick value by 1000

ticks = ax.get_yticks()             # grab the current tick *positions*
ax.set_yticks(ticks)                # (not strictly necessary if you’re keeping them)
ax.set_yticklabels(ticks / 1000)    # divide each tick value by 1000

scalebar_main = AnchoredSizeBar(
    ax.transData,             # data‐coords transform
    bar_len_parent,           # length in µas
    f"{-bar_len_parent/1000:.0f} mas",
    loc="lower left",         # pick a corner
    pad=0.8, borderpad=0.5,
    color="black",
    frameon=False,
    size_vertical=0,
    sep=15,                   # separation from the axes
)
ax.add_artist(scalebar_main)


axins = inset_axes(ax,
                   width="72.5%",   # % of parent axes
                   height="72.5%",
                   loc=4, borderpad=0,bbox_to_anchor=(+0.025,-.0, 1,0.99),
    bbox_transform=ax.transAxes,)

# redraw both datasets in the inset
axins.imshow(data86, origin="lower", cmap="gray_r", extent=ext86, interpolation="gaussian", vmin=11, vmax=12)
axins.contour(data230, levels=[1e-4,3e-4,6e-4],
              colors="black", linewidths=1.5,
              origin="lower", extent=ext230)



# zoom to the cutout regio

x1, x2 = 1930, 2080
y1, y2 =  1800, 2000

x2_p, y2_p = 72, 103

x2_o, y2_o = 60, 86


x_inset = x2-x2_p
y_inset = y2-y2_p

x_inset_o = x2-x2_o
y_inset_o = y2-y2_o

axins.scatter(x_inset, y_inset, s=300, color='red', marker='x', zorder=11, linewidths=5)
axins.scatter(x_inset_o, y_inset_o, s=300, color='red', marker='x', zorder=11, linewidths=5)


xs = np.array([x_inset, x_inset_o])
ys = np.array([y_inset, y_inset_o])
m, b = np.polyfit(xs, ys, 1)

x_line = np.linspace(x1, x2, 100)
y_line = m*x_line + b
axins.plot(x_line, y_line, color="red", linestyle="-", linewidth=1,
           label=f"Core shift", zorder=12)
axins.scatter(0, 0, s=0, color='white', marker='x', zorder=11, linewidths=5, label="Uncertainties:")

axins.set_xlim(x1, x2)
axins.set_ylim(y1, y2)
axins.set_xticks([])
axins.set_yticks([])

scalebar_inset = AnchoredSizeBar(
    axins.transData,
    bar_len_inset,
    f"{-bar_len_inset:.0f} µas",
    loc="lower left",
    pad=0.8, borderpad=0.3,
    color="black",
    frameon=False,
    size_vertical=0,
    sep=15,
)
axins.add_artist(scalebar_inset)

x0, y0   = x_inset_o, y_inset_o    # center of the ellipse
width    = error_propagation_3*1000  * 2    # full width in µas
height   = error_propagation_3*1000  * 2         # full height in µas
angle    = 30            # rotation in degrees (counterclockwise)
# 2. Create the Ellipse patch:
ellipse_3 = Ellipse(
    xy=(x0, y0),         # center
    width=width,
    height=height,
    angle=angle,
    facecolor="lightgreen",    # transparent fill
    alpha=0.5,          # transparency
    zorder=10,
    label=f"Single scan, 3 antennas \n ignoring source structure"  # label for legend
)

ellipse_3_nosource = Ellipse(
    xy=(x0, y0),         # center
    width=error_propagation_3_without_source*1000 * 2,
    height=error_propagation_3_without_source*1000 * 2,
    angle=angle,
    facecolor="orange",    # transparent fill
    alpha=0.8,          # transparency
    zorder=14,
    label=f"Single scan, 7 antennas"  # label for legend
)

ellipse_best_case = Ellipse(
    xy=(x0, y0),         # center
    width=error_propagation_all_scans_tennas_no_source*1000 * 2,
    height=error_propagation_all_scans_tennas_no_source*1000 * 2,
    angle=angle,
    facecolor="blue",    # transparent fill
    alpha=0.8,          # transparency
    zorder=15,
    label=f"All day, 7 antennas"  # label for legend
)

# 3. Add it to your axes:
axins.add_patch(ellipse_3)
axins.add_patch(ellipse_3_nosource)
axins.add_patch(ellipse_best_case)
# draw the box on the main axes
mark_inset(ax, axins, loc1=2, loc2=2, fc="none", ec="black", lw=0)

axins.invert_xaxis()  # invert x-axis for the inset

# grab everything from the inset legend
h, l = axins.get_legend_handles_labels()

# draw the same legend on the parent axes
ax.legend(h, l, loc='upper right', fontsize=30)

# plt.tight_layout()
# plt.show()
```
